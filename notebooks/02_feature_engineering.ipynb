Extract Hold Time (`H.*`) and Flight Time (`DD.*`) vectors and build subject-wise enrollment / verification splits.

```python
import sys
from pathlib import Path
import numpy as np

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))

from src.data.loader import load_dataset, list_subjects
from src.data.splits import build_subject_split, build_training_dataset
from src.features.keystroke import extract_features, get_feature_columns, list_feature_sets

df = load_dataset()
print(list_feature_sets())
for name in list_feature_sets():
    X = extract_features(df.head(5), name)
    print(name, X.shape)
```
